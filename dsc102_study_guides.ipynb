9. A programmer writes two loops to sum a 1000x1000 integer matrix stored in row-major
order in memory: Version A: for i in rows: for j in cols: total += M[i][j] Version B: for j in
cols: for i in rows: total += M[i][j]
a. Which version runs faster on a modern CPU? Explain why in terms of cache lines
and spatial locality. A
b. What cache behavior does the slower version exhibit? Cache Thrashing
Answer: Version A, cache thrashing
Row-major storage means M[i][0], M[i][1]... are contiguous. Version A walks sequentially

causing a cache miss every access. "Cache thrashing" is the right term here.
